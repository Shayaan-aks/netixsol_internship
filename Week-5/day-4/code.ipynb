## Task 1: Multi-Agent Design Thinking & Role Decomposition

### Business Scenario
**Competitive Intelligence & Strategic Product Positioning** for AI Coding Assistants.

### Agent Role Decomposition
1. **Senior Competitive Intelligence Analyst (`researcher`)**
   - **Role:** Market Intelligence Specialist
   - **Goal:** Gather technical specs, pricing models, feature matrices, and market movements for competitors.
   - **Backstory:** Deep software architecture background; evaluates tech specs objectively without marketing bias.
2. **Product Strategy & Data Analyst (`analyst`)**
   - **Role:** Data-Driven Product Strategist
   - **Goal:** Synthesize raw research notes into quantitative feature gap matrices and SWOT scores.
   - **Backstory:** Enterprise product background; transforms qualitative notes into structured decision tables.
3. **Principal Marketing Strategist & Copywriter (`copywriter`)**
   - **Role:** B2B Tech Marketing Director
   - **Goal:** Convert SWOT matrices into executive value propositions, positioning hooks, and go-to-market briefs.
   - **Backstory:** Expert in translating technical feature advantages into compelling executive narratives.

### Why Multi-Agent Specialized Crews Outperform Generalist Agents
Specialized agents outperform single generalists by constraining context windows, enforcing strict role personas, and preventing cognitive overload. A single generalist forced to research, analyze, and format simultaneously often produces shallow analysis or loses tone consistency. However, for simple single-turn tasks, single agents avoid multi-agent delegation latency and overhead cost.
