```python
import pandas as pd

import kagglehub

# Download latest version
path = kagglehub.dataset_download("brendan45774/test-file")

print("Path to dataset files:", path)
```

```python
import os

# Join the folder path with the file name
df = pd.read_csv(os.path.join(path, "tested.csv"))
df
```

```python
df[pd.isnull(df.Age)]
```

```python
df['Age']=df['Age'].fillna(df.groupby(['Sex','Pclass'])['Age'].transform('median'))
```

i have mitigated the null age values based on median of age grouped based on Sex and Pclass because this seems like the best option available at the time

```python
df.dtypes
```

```python
df['Survived']=df['Survived'].astype(bool)
```

changed survived from int to boolean


```python
df.drop_duplicates(subset=['Name'],inplace=True)
```

duplicates dropped based on name 


```python
def age_group(age):
    if age <= 12:
        return "Child"
    elif age <= 19:
        return "Teen"
    elif age <= 35:
        return "Young Adult"
    elif age <= 60:
        return "Adult"
    else:
        return "Senior"

df['Age_Group'] = df['Age'].apply(age_group)
```

new column age group created 

```python
df['Family size']=df['SibSp']+df['Parch']+1
```

new column family size created 

```python
df.groupby(['Sex']).Age.agg([len,min,max])
```

```python
class_info = pd.DataFrame({
    'Pclass': [1, 2, 3],
    'Class_Name': ['First Class', 'Second Class', 'Third Class'],
    'Luxury_Level': ['Luxury', 'Standard', 'Economy']
})
```

```python
df=df.merge(class_info,on='Pclass',how='left')
```

merged with a new class info dataset for more context
