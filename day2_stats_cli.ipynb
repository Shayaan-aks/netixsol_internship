```python
import numpy as np
```

```python
rng = np.random.default_rng()
```

```python

array=rng.normal(loc=10, scale=30, size=(20,20))
array
```

```python
mean = array.mean()
minimum = array.min()
maximum = array.max()
std_dev = array.std()

print(f"Mean    : {mean:.4f}")
print(f"Min     : {minimum:.4f}")
print(f"Max     : {maximum:.4f}")
print(f"Std Dev : {std_dev:.4f}")
```

```python
from numpy.lib.stride_tricks import sliding_window_view
windows=sliding_window_view(array,(3,3)) 

for i in range(windows.shape[0]):
    for j in range(windows.shape[1]):
        window=windows[i,j]
        print(
            f"mean{window.mean():.2f}"
            f"standard deviation{window.std():.2f}"
        )
```

```python
#z-score normalized
normalized_array=(array-array.mean())/array.std()
normalized_array
```

```python
testarray=normalized_array.flatten()
outliers=[]
for i in range(len(testarray)):
    if testarray[i]>2:
        outliers.append(testarray[i])
print(len(outliers))
print(outliers)
```
